2.	Explain the difference between List and Tuple in Python.
Ans. In Python, lists and tuples both store collections of data, but differ in mutability, performance and memory usage. A list is mutable, meaning you can change, add, or remove its elements after creation. In contrast, a tuple is immutable; once it is defined, its contents and size cannot be altered.

3.	What is a function in Python? Why are functions useful?
Ans: A function is a reusable block of code that performs a specific task or set of tasks that only returns when it is called. If there is a certain code that must be written repeatedly for different inputs, we make use of functions.
Functions are useful because:
•	It is easy to read and understand.
•	We can reuse the code multiple times.
•	It is easier to organize the code.
•	Complex code can be broken into simpler code.
